```python
#nested conditional statements
#even, odd or negative 
num=int(input("enter a number: "))
if num>0:
    print("positive number")
    if num%2==0:
        print("even number")
    else:
        print("odd number")
else:
    print("number is zero or negative")     
```

```python
#leap year nested if example

yr=int(input("enter the year: "))
if (yr%4==0):
    if yr%100==0:
        if yr%400==0:
            print( yr, "is a leap year")
        else:
            print("this year is not a leap year!")
    else:
        print(yr, " is a leap year")
else: 
    print(yr, "is not a leap year")
```

```python
#Simple caluclator using elif 

a = float(input("1st number:  "))
b = float(input("2nd number:  "))
ch= input("enter operator (+, -, *, /)")

if ch== '+':
    result= a+b
elif ch== '-':
    result= a-b
elif ch== '*':
    result= a*b
elif ch== '/':
    if b!=0:
        result= a/b
    else:
        result = "error! Division by zero"
else:
    result= "invalied input"
print("result:", result)
```

```python
#ticket pricing based on age and student status
#nested if
age= int(input("enter age: "))
std= input("are you a student? (y/n): ").lower()

if age < 5:
    price = "Free"
elif age <= 12:
    price = "$10"
elif age <= 17:
    if std == 'yes':
        price = "$12"
    else:
        price = "$15"
elif age <= 64:
    if std == 'yes':
        price = "$18"
    else:
        price = "$25"
else:
    price = "$20"

print("Ticket Price:", price)
```

loop control statements (brake,continue,pass)

```python
#brake
for i in range(10):
    if i==6:
        break
    print(i)
```

```python
#continue
for i in range(10):
    if i%2==0:
        continue
    print(i)
```

```python
#pass (does nothing)
for i in range(5):
    if i==3:
        print("next no is ",i)
        pass
    print(i)
```

```python
# Nested loops
for i in range(3):
    for j in range(2):
        print(f"i={i} and j= {j}")
```

```python
#example sum of 1st N natural nos with while loop

n=10
sum=0
count=1
while count<=n:
    sum=sum+count
    count+=1
print("sum = ",sum)
```

```python
#example sum of 1st N natural nos with for loop

n=10
sum=0
for i in range(n+1):
    sum=sum+i
print("sum = ",sum)
```

```python
#example prime nos btw 1-100
for num in range(1,100):
    if num>1:
        for i in range(2,num):
            if num%i==0:
                break
        else:
            print(num)
```
